```python
import pandas as pd
import matplotlib.pyplot as plt

# Data from the spreadsheet
data = {
    'Group': ['a120', 'b100', 'nt', 'met', 'nor'],
    'ini': [319.33, 254, 256.3, 330, 102.6],
    'wk1': [196.33, 79.66, 298, 234, 97],
    'wk2': [72, 75.66, 302, 250, 100],
    'wk3': [64, 106.33, 370, 225, 93]
}

# Create DataFrame
df = pd.DataFrame(data)

# Plot
plt.figure(figsize=(10, 6))
for group in df['Group']:
    row = df[df['Group'] == group]
    plt.plot(row.columns[1:], row.values[0][1:], marker='o', label=group)

plt.title('Glucose Levels Over Weeks')
plt.xlabel('Week')
plt.ylabel('Glucose Level')
plt.legend(title='Group')
plt.grid(True)
plt.savefig("glucosweeks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Data from the spreadsheet
data = {
    'MINUTES': [0, 15, 30, 60, 90, 120],
    'A120': [97, 87, 85, 81, 79, 76],
    'B100': [118, 111, 117, 95, 95, 91],
    'NT': [175, 179, 200, 203, 210, 210],
    'MET.': [145, 139, 133, 128, 124, 121],
    'NOR': [115, 113, 113, 109, 107, 101]
}

# Create DataFrame
df = pd.DataFrame(data)

# Plot
plt.figure(figsize=(10, 6))
for column in df.columns[1:]:
    plt.plot(df['MINUTES'], df[column], marker='o', label=column)

plt.title('Glucose Levels Over Time (Minutes)')
plt.xlabel('Time (minutes)')
plt.ylabel('Glucose Level (mg/dl)')
plt.legend(title='Group')
plt.grid(True)

# Set x-axis ticks to show all time points
plt.xticks(df['MINUTES'])
plt.savefig("glucosminut")
plt.show()
```

```python
food = input ('Please enter the food name')
if food in indian:
    print ('food is indian')
elif food in italy:
    print ('food is italian')
elif food in American:
    print ('food is American')
elif food in Nigeria:
    print ('food is Nigerian')
else:
    print ('I dont know which food is this')
```
